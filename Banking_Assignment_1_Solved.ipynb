challenge

| Action | Why it is dangerous |
|---|---|
| `collect()` | Pulls **all** rows to the driver. Billions of transactions cause driver out-of-memory crashes. Sensitive data also ends up in one place. |
| `toPandas()` | Same problem as `collect()`: everything lands in driver memory. |
| `count()` | Not memory-risky, but scans the entire dataset; it can be slow and costly on huge data. |
| `countByValue()` | Returns a dict to the driver. With high-cardinality columns (e.g. `customer_id`) the dict is huge and can crash the driver. |
| `foreach()` | Runs on executors, so `print` output is not visible on the driver; slow row-by-row side effects (e.g. calling an external API per row) can overload systems. |
| `toLocalIterator()` | Safer than `collect()` (one partition at a time) but still slow, since it runs jobs partition by partition. |
| `write` | Can produce huge or too many small files, and `overwrite` mode can wipe existing data. |
| `show()` / `take()` / `head()` / `first()` / `isEmpty()` | Generally safe; they fetch only a few rows. |
